```python
from transformers import pipeline
```

```python
summarizer = pipeline(
    "text-generation",
    model="sshleifer/distilbart-cnn-12-6"
)
```

```python
text = """
Artificial Intelligence is transforming many industries.
In healthcare, AI helps doctors analyze medical images,
detect diseases, and support treatment decisions.
In education, AI enables personalized learning experiences.
Businesses use AI to automate repetitive tasks and
improve decision-making. However, AI also introduces
challenges such as data privacy, bias, and job displacement.
Responsible development is important to ensure that
AI benefits society.
"""
```

```python
summary = summarizer(
    text,
    max_length=60,
    min_length=20,
    do_sample=False
)

print(summary[0]["summary_text"])
```

```python
print("ORIGINAL TEXT:")
print(text)

print("\nGENERATED SUMMARY:")
print(summary[0]["generated_text"])
```
